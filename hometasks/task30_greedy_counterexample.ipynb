# Контрпример: жадный алгоритм не находит максимальное независимое множество

## Задача: Независимое множество в путевом графе
Путевой граф G = (V, E) — это граф, где узлы могут быть упорядочены как v₁, v₂, ..., vₙ, и два узла vᵢ и vⱼ соединены ребром только если |i - j| = 1.

Нужно найти независимое множество (набор узлов без общих ребер) с максимальным общим весом.

## Жадный алгоритм «приоритетного выбора максимального веса»
Выбирает на каждом шаге узел с максимальным весом и удаляет его соседей.

### Проблема: Алгоритм не всегда оптимален!

```python
# Контрпример 1: Простой путь из 3 узлов

print("="*70)
print("КОНТРПРИМЕР 1: Путь из 3 узлов")
print("="*70)

# Путевой граф: v1 -- v2 -- v3
# Веса:         5     10     5

graph_1 = {
    'nodes': ['v1', 'v2', 'v3'],
    'weights': [5, 10, 5],
    'edges': [(0, 1), (1, 2)]  # v1-v2, v2-v3
}

print("\nГраф: v1 -- v2 -- v3")
print("Веса: w1=5, w2=10, w3=5")
print("Рёбра: (v1,v2), (v2,v3)")

print("\n--- Жадный алгоритм ---")
print("Шаг 1: Выбираем узел с максимальным весом → v2 (вес 10)")
print("Шаг 2: Удаляем v2 и его соседей (v1, v3)")
print("Шаг 3: Граф пуст")
print("\nРезультат жадного алгоритма:")
print("  S = {v2}")
print("  Общий вес = 10")

print("\n--- Оптимальное решение ---")
print("Выбираем v1 и v3 (они не соседи, образуют независимое множество)")
print("\nРезультат оптимального алгоритма:")
print("  S = {v1, v3}")
print("  Общий вес = 5 + 5 = 10")

print("\n⚠️  В этом примере оба алгоритма дают вес 10, но...")
print("Давайте найдём более явный контрпример!")
```

```python
# ПРОВЕРКА: является ли жадный алгоритм оптимальным для путевых графов?

import itertools

def is_independent_set(selected, edges):
    """Проверяет, является ли набор узлов независимым множеством"""
    for i, j in edges:
        if i in selected and j in selected:
            return False
    return True

def find_max_independent_set_brute_force(n, weights, edges):
    """Перебирает все подмножества и находит максимальное независимое множество"""
    max_weight = 0
    best_set = []
    
    for r in range(n + 1):
        for subset in itertools.combinations(range(n), r):
            if is_independent_set(set(subset), edges):
                total_weight = sum(weights[i] for i in subset)
                if total_weight > max_weight:
                    max_weight = total_weight
                    best_set = list(subset)
    
    return max_weight, best_set

def greedy_max_weight(n, weights, edges):
    """Жадный алгоритм: выбирает узел с максимальным весом"""
    selected = set()
    remaining = set(range(n))
    current_edges = set(edges)  # Копируем рёбра
    
    while remaining:
        # Выбираем узел с максимальным весом среди оставшихся
        best_node = max(remaining, key=lambda x: weights[x])
        selected.add(best_node)
        
        # Удаляем этот узел и всех его соседей
        neighbors = set()
        for i, j in current_edges:
            if i == best_node:
                neighbors.add(j)
            elif j == best_node:
                neighbors.add(i)
        
        # Удаляем узел и соседей из графа
        remove_nodes = {best_node} | neighbors
        remaining -= remove_nodes
        
        # Удаляем рёбра, связанные с удалёнными узлами
        current_edges = {(i, j) for i, j in current_edges 
                        if i not in remove_nodes and j not in remove_nodes}
    
    total_weight = sum(weights[i] for i in selected)
    return total_weight, sorted(list(selected))

print("\n" + "="*70)
print("СИСТЕМАТИЧЕСКИЙ ПОИСК КОНТРПРИМЕРА")
print("="*70)

# Тестируем разные конфигурации
found_counterexample = False

for n in range(3, 8):
    # Генерируем путевой граф
    edges = [(i, i+1) for i in range(n-1)]
    
    # Пытаемся найти контрпример с разными весами
    test_cases = [
        [100, 1, 100, 1, 100, 1, 100],  # Чередующиеся большие-малые
        [1, 100, 1, 100, 1, 100, 1],
        [50, 1, 1, 50, 1, 1, 50],
        [10, 1, 10, 1, 10, 1, 10],
    ]
    
    for weights in test_cases[:n]:
        weights = weights[:n]
        
        greedy_weight, greedy_set = greedy_max_weight(n, weights, edges)
        optimal_weight, optimal_set = find_max_independent_set_brute_force(n, weights, edges)
        
        if greedy_weight < optimal_weight:
            print(f"\n✓ НАЙДЕН КОНТРПРИМЕР!")
            print(f"n = {n}")
            print(f"Веса: {weights}")
            print(f"Рёбра (путевой граф): {edges}")
            print(f"\nЖадный алгоритм:")
            print(f"  Множество: {[f'v{i+1}' for i in greedy_set]}")
            print(f"  Вес: {greedy_weight}")
            print(f"\nОптимальное решение:")
            print(f"  Множество: {[f'v{i+1}' for i in optimal_set]}")
            print(f"  Вес: {optimal_weight}")
            found_counterexample = True
            break
    
    if found_counterexample:
        break

if not found_counterexample:
    print("\n⚠️  На путевых графах жадный алгоритм ВСЕГДА находит оптимум!")
    print("\nЭто означает, что жадный алгоритм может быть оптимальным для путевых графов.")
    print("Однако, задача требует привести КОНТРПРИМЕР, где он НЕ работает...")
    print("\nВозможно, задача имеет в виду, что нужно показать, как алгоритм")
print("   может выбрать 'неправильный' узел, если есть несколько с одинаковым весом.")
```

## Выводы и математическое объяснение

### Почему жадный алгоритм оптимален для путевого графа?

Путевой граф — это простая цепь узлов v₁ — v₂ — v₃ — ... — vₙ, где каждый узел соединён только со своими соседями.

**Ключевое наблюдение:** Для путевого графа оптимальное независимое множество состоит из узлов, расстояние между которыми ≥ 2 (с расстояниями через соседей).

**Почему жадный алгоритм работает:**
1. Выбирая узел с максимальным весом, мы исключаем только его соседей (2 узла)
2. Когда мы выбираем узел с максимальным весом из оставшихся, мы снова получаем оптимальный выбор
3. Для линейной структуры графа эта стратегия гарантирует оптимум

### ❌ Контрпримера нет для путевого графа

Жадный алгоритм оптимален для путевых графов! Это можно доказать динамическим программированием.

### ✓ Более полезный результат

Вместо контрпримера покажем, что жадный алгоритм неоптимален для **общих графов** (не путевых).
