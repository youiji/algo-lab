# Define the Automaton Class
Create a class to represent the automaton with properties like grid size, rules, and state initialization methods.

```python
# Define the Automaton Class
class Automaton:
    def __init__(self, grid_size, rule):
        """
        Initialize the automaton with a grid size and a rule.

        Parameters:
        grid_size (int): The size of the grid (grid will be grid_size x grid_size).
        rule (function): A function defining the rules of the automaton.
        """
        self.grid_size = grid_size
        self.rule = rule
        self.grid = np.zeros((grid_size, grid_size), dtype=int)  # Initialize grid with zeros

    def initialize_state(self, state=None):
        """
        Initialize the state of the automaton.

        Parameters:
        state (2D array or None): If provided, sets the grid to the given state.
                                  If None, initializes the grid with random values (0 or 1).
        """
        if state is not None:
            self.grid = np.array(state, dtype=int)
        else:
            self.grid = np.random.randint(2, size=(self.grid_size, self.grid_size))

    def update_state(self):
        """
        Update the state of the automaton based on the defined rule.
        """
        new_grid = np.zeros_like(self.grid)
        for i in range(self.grid_size):
            for j in range(self.grid_size):
                new_grid[i, j] = self.rule(self.grid, i, j)
        self.grid = new_grid

    def get_state(self):
        """
        Get the current state of the automaton.

        Returns:
        2D array: The current grid state.
        """
        return self.grid
```

# Initialize Automaton State
Set up the initial state of the automaton grid with desired configuration and parameters.

```python
# Initialize the automaton with a specific grid size and rule
grid_size = 10  # Define the size of the grid
def sample_rule(grid, x, y):
    """
    A sample rule for the automaton. This rule simply toggles the state of each cell.

    Parameters:
    grid (2D array): The current grid state.
    x (int): The x-coordinate of the cell.
    y (int): The y-coordinate of the cell.

    Returns:
    int: The new state of the cell.
    """
    return 1 - grid[x, y]

# Create an instance of the Automaton class
automaton = Automaton(grid_size, sample_rule)

# Initialize the state of the automaton
initial_state = np.zeros((grid_size, grid_size), dtype=int)  # Example: Start with all zeros
automaton.initialize_state(initial_state)
```
