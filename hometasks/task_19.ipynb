```python
def min_base_stations(houses):
    houses.sort()
    n = len(houses)
    i = 0

    stations = 0
    positions = []

    while i < n:
        left = houses[i]
        station_pos = left + 4
        cover_end = left + 8
        stations += 1
        positions.append(station_pos)

        while i < n and houses[i] <= cover_end:
            i += 1

    return stations, positions
```

```python
min_base_stations([1, 2, 3, 10, 11, 12, 20])
```
