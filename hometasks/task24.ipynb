```python
from collections import defaultdict

edges = [
    (0, 1, 4),
    (0, 2, 3),
    (1, 2, 1),
    (1, 3, 2),
    (2, 3, 4),
    (3, 4, 2),
    (4, 5, 6),
]

V = 6


def kruskal(edges, V):
    p = list(range(V))

    def find(x):
        while p[x] != x:
            x = p[x]
        return x

    mst = []

    for u, v, w in sorted(edges, key=lambda x: x[2]):
        ru, rv = find(u), find(v)

        if ru != rv:
            p[ru] = rv
            mst.append((u, v, w))

    return mst


def update_mst(mst, edge):
    u, v, w = edge

    g = defaultdict(list)

    for i, (a, b, c) in enumerate(mst):
        g[a].append((b, c, i))
        g[b].append((a, c, i))

    stack = [(u, -1)]
    parent = {}

    while stack:
        x, prev = stack.pop()

        if x == v:
            break

        for to, ww, idx in g[x]:
            if to != prev:
                parent[to] = (x, ww, idx)
                stack.append((to, x))

    mx = -1
    pos = -1
    cur = v

    while cur != u:
        prev, ww, idx = parent[cur]

        if ww > mx:
            mx = ww
            pos = idx

        cur = prev

    if w < mx:
        mst[pos] = edge

    return mst


mst = kruskal(edges, V)

print("MST:")
print(*mst, sep="\n")

new_edge = (2, 5, 1)

print("\nДобавляем:", new_edge)

mst = update_mst(mst, new_edge)

print("\nНовое MST:")
print(*mst, sep="\n")
```
