```python
import sys
from collections import deque

content = sys.stdin.readlines()
num_trucks, num_hours = map(int, content[0].strip().split())
schedules = [list(map(int, line.strip().split())) for line in content[1:num_trucks+1]]

arrival_time = [[-1] * num_trucks for _ in range(num_trucks)]
preferences = [[] for _ in range(num_trucks)]

for truck_id in range(num_trucks):
    for hour, warehouse_num in enumerate(schedules[truck_id]):
        if warehouse_num != 0:
            warehouse_index = warehouse_num - 1
            arrival_time[truck_id][warehouse_index] = hour
            preferences[truck_id].append(warehouse_index)

next_choice = [0] * num_trucks
current_occupant = [-1] * num_trucks
free_trucks = deque(range(num_trucks))

while free_trucks:
    truck = free_trucks.popleft()
    warehouse = preferences[truck][next_choice[truck]]
    next_choice[truck] += 1
    occupant = current_occupant[warehouse]

    if occupant == -1:
        current_occupant[warehouse] = truck
    elif arrival_time[truck][warehouse] > arrival_time[occupant][warehouse]:
        current_occupant[warehouse] = truck
        free_trucks.append(occupant)
    else:
        free_trucks.append(truck)

final_warehouse = [0] * num_trucks
for warehouse_index in range(num_trucks):
    truck = current_occupant[warehouse_index]
    final_warehouse[truck] = warehouse_index + 1

for w in final_warehouse:
    print(w)
```
