```python
def solve(r):
    n = len(r)
    size = 1 << n

    dp = [-1.0] * size
    dp[size - 1] = 0

    def get(mask):
        if dp[mask] != -1:
            return dp[mask]

        t = bin(mask).count("1") + 1

        best = 0
        for i in range(n):
            if not (mask & (1 << i)):
                best = max(best, 100 * (r[i] ** t) + get(mask | (1 << i)))

        dp[mask] = best
        return best

    total = get(0)

    mask = 0
    order = []

    for t in range(1, n + 1):
        best_i = -1
        best_val = -1

        for i in range(n):
            if not (mask & (1 << i)):
                val = 100 * (r[i] ** t) + get(mask | (1 << i))
                if val > best_val:
                    best_val = val
                    best_i = i

        order.append(r[best_i])
        mask |= (1 << best_i)

    return total, order
```

```python
r = [0.75, 0.5, 0.01]

print(solve(r))
```
