```python
from typing import List, Tuple

def max_weight_independent_set(weights: List[int]) -> int:

    n = len(weights)
    
    if n == 0:
        return 0
    if n == 1:
        return weights[0]
    
    dp = [0] * n
    dp[0] = weights[0]  
    dp[1] = max(weights[0], weights[1])  
    
    for i in range(2, n):
        dp[i] = max(dp[i-1], weights[i] + dp[i-2])
    
    return dp[n - 1]



weights1 = [5, 1, 3, 9, 4]
result1 = max_weight_independent_set(weights1)
print(f"\nТест 1: weights = {weights1}")
print(f"Результат: {result1}")


weights2 = [10]
result2 = max_weight_independent_set(weights2)
print(f"\nТест 2: weights = {weights2}")
print(f"Результат: {result2}")

# Тест 3: Две вершины
weights3 = [3, 5]
result3 = max_weight_independent_set(weights3)
print(f"\nТест 3: weights = {weights3}")
print(f"Результат: {result3}")

```
