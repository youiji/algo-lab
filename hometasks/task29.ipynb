```python
import bisect
from typing import List, Tuple

def max_billboard_profit(M: int, positions: List[float], profits: List[int]) -> int: 
    # Обработка граничного случая
    if not positions:
        return 0
    
    n = len(positions)
    bills = [(positions[i], profits[i]) for i in range(n)]
    bills.sort()
    
    sorted_x = [bill[0] for bill in bills]
    sorted_r = [bill[1] for bill in bills]
    
    dp = [0] * n
    dp[0] = sorted_r[0]  
    
    for i in range(1, n):
        profit_without = dp[i - 1]
        
        current_x = sorted_x[i]
        current_r = sorted_r[i]
        threshold = current_x - 5
        j = bisect.bisect_left(sorted_x, threshold) - 1
        
        if j >= 0:
            profit_with = current_r + dp[j]
        else:
            profit_with = current_r
        
        dp[i] = max(profit_without, profit_with)
    
    return dp[n - 1]


M = 20
positions = [6, 7, 12, 14]
profits = [5, 6, 5, 1]

result = max_billboard_profit(M, positions, profits)
print(f"M = {M}, позиции = {positions}, прибыли = {profits}")
print(f"Максимальная прибыль = {result}")
```
