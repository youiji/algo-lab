```python
def is_subsequence(sequence, pattern):
    j = 0  

    for event in sequence:
        if j < len(pattern) and event == pattern[j]:
            j += 1

    return j == len(pattern)
```

```python
S = [
    "купить Apple",
    "купить Google",
    "купить Tesla",
    "купить Amazon",
    "купить Microsoft"
]

pattern = [
    "купить Google",
    "купить Amazon"
]

is_subsequence(S, pattern)
```

```python
pattern = [
    "купить Amazon",
    "купить Google"
]

is_subsequence(S, pattern)
```
