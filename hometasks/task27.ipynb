```python
def find_peak(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        mid = (left + right) // 2

        if arr[mid] < arr[mid + 1]:
            left = mid + 1
        else:
            right = mid

    return left


tests = [
    ([1, 3, 7, 12, 9, 5, 2], 3),
    ([1, 2, 3, 4, 5], 4),
    ([5, 4, 3, 2, 1], 0),
    ([1, 5, 10, 7, 3], 2),
]

for arr, expected in tests:
    result = find_peak(arr)

    print(f"array      = {arr}")
    print(f"peak index = {result}")
    print(f"peak value = {arr[result]}")
    print(f"expected   = {expected}")
    print('-'*40)

```
