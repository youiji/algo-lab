```python
def best_trade(prices):
  
    n = len(prices)

    if n < 2:
        return None

    min_price = prices[0]
    min_day = 1

    best_profit = 0
    best_buy = None
    best_sell = None

    for day in range(2, n + 1):
        current_price = prices[day - 1]

        current_profit = current_price - min_price

        if current_profit > best_profit:
            best_profit = current_profit
            best_buy = min_day
            best_sell = day

        if current_price < min_price:
            min_price = current_price
            min_day = day

    if best_profit <= 0:
        return None

    return best_buy, best_sell, best_profit
```

```python
tests = [
    [9, 1, 5],
    [5, 4, 3, 2],
    [1, 2, 3, 4, 5],
    [7, 1, 5, 3, 6, 4],
    [2, 10, 1, 20],
]

for prices in tests:
    result = best_trade(prices)

    print(f"prices = {prices}")

    if result is None:
        print("Прибыль невозможна")
    else:
        buy, sell, profit = result

        print(f"Покупать: день {buy}")
        print(f"Продавать: день {sell}")
        print(f"Прибыль: {profit}")

    print("-" * 40)
```
