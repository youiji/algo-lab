```python
def triathlon_schedule(data):

    participants = [
        (i + 1, s, b + r)
        for i, (s, b, r) in enumerate(data)
    ]
    print(participants)
    participants.sort(key=lambda x: x[2], reverse=True)

    time_swim = 0
    order = []
    finish_times = []
    for idx, swim, qr in participants:
        time_swim += swim
        finish = time_swim + qr

        order.append(idx)
        finish_times.append((idx, finish))

    makespan = max(f for _, f in finish_times)

    print("Порядок участников:", order)

    print("\nВремя завершения:")
    for i, t in finish_times:
        print(f"Участник {i}: {t}")

    print("\nОбщее время:", makespan)


info = [
        (10, 30, 20),
        (12, 25, 15),
        (8, 40, 10),
        (15, 20, 25),
    ]

triathlon_schedule(info)
```
