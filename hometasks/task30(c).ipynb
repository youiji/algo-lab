```python
def max_weight_independent_set(weights):
    n = len(weights)
    if n == 0:
        return 0, []
    if n == 1:
        return weights[0], [1]

    dp = [0] * (n + 1)
    dp[1] = weights[0]

    for i in range(2, n + 1):
        dp[i] = max(dp[i-1], dp[i-2] + weights[i-1])

    max_weight = dp[n]
    independent_set = []
    i = n
    while i >= 1:
        if dp[i] == dp[i-1]:
            i -= 1
        else:
            independent_set.append(i)  
            i -= 2

    return max_weight, independent_set[::-1]

weights = [10, 1, 1, 10]
weight, nodes = max_weight_independent_set(weights)
print(f"Максимальный вес: {weight}")
print(f"Выбранные вершины (номера): {nodes}")
```
